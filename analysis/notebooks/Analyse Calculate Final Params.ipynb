Calculate centrality values for above nodes and arrange in dataframe

```python
G = nx.from_numpy_array(utils.WHOLE_BRAIN_CONN)
bet_cent = nx.betweenness_centrality(G)
deg_cent = nx.degree_centrality(G)
eig_cent = nx.eigenvector_centrality(G)
efficiency = {}
for node in G.nodes:
    # Calculate the sum of reciprocal of shortest path lengths from the node to all other nodes
    sp_lengths = nx.shortest_path_length(G, source=node)
    sum_reciprocal_sp = sum(1 / length for length in sp_lengths.values() if length > 0)

    # Calculate efficiency: E(v) = (1 / (N-1)) * sum(1 / d(v, u))
    # where N is the number of nodes and d(v, u) is the shortest path length between v and u
    efficiency[node] = sum_reciprocal_sp / (len(G.nodes) - 1)


bc_vals = []
deg_vals = []
ev_vals = []
ne_vals = []

for i in nodes:
    bc_vals.append(bet_cent[i])
    deg_vals.append(deg_cent[i])
    ev_vals.append(eig_cent[i])
    ne_vals.append(efficiency[i])

print(f"Betweenness centrality values: {bc_vals}")
print(f"Degree centrality values: {deg_vals}")
print(f"Eigen vector centrality values: {ev_vals}")
print(f"Nodal Efficiency values: {ne_vals}")
```

```python
# removing duplicate data from centrality values so that length matches
def dup_cent_rem(cent):
    j = 0
    val = []
    while j < len(cent):
        val.append(cent[j])
        j += 2

    return val

# z scoring all values
b_d = dup_cent_rem(bc_vals)
b = [(x - np.mean(b_d)) / np.std(b_d) for x in b_d]
b.append(0)

n_d = dup_cent_rem(ne_vals)
n = [(x - np.mean(n_d)) / np.std(n_d) for x in n_d]
n.append(0)

d_d = dup_cent_rem(deg_vals)
d = [(x - np.mean(d_d)) / np.std(d_d) for x in d_d]
d.append(0)

e_d = dup_cent_rem(ev_vals)
e = [(x - np.mean(e_d)) / np.std(e_d) for x in e_d]
e.append(0)

df_cent = pd.DataFrame(
    {"BET_CENT": b, "NOD_EFF": n, "DEG_CENT": d, "EIG_CENT": e}, index=node_label
)

df_bc = df_cent.sort_values("BET_CENT", ascending=True, inplace=False)
df_bc = df_bc.drop("fn")
print(df_bc)
```

```python
# defining dictionaries for each parameter
# storing both partial and full transition data
# store standard deviation and manWhitney U test values

num_trans = {}
n_p = []
n_p_s = []
n_f = []
n_f_s = []

trans_time = {}
t_p = []
t_p_s = []
t_f = []
t_f_s = []
t_p_mw = []
t_f_mw = []

fp_time = {}
f_p = []
f_p_s = []
f_f = []
f_f_s = []
f_p_mw = []
f_f_mw = []

sync_time = {}
s_p = []
s_p_s = []
s_f = []
s_f_s = []
s_p_mw = []
s_f_mw = []

cov = {}
cov_p = []
cov_p_s = []
cov_f = []
cov_f_s = []
cov_p_mw = []
cov_f_mw = []

# node_label=['PERI','ENTI','ORB','RE','BLA','PIR','MOp','MOs','LHA','MOB','LSc','MRN']
# getting networks in ascending order
network_asc = [
    "fn",
    "mob",
    "lsc",
    "orb",
    "pir",
    "bla",
    "mrn",
    "re",
    "enti",
    "peri",
    "mos",
    "lha",
    "mop",
]
fn_data = reader.data_read_pkl("results/calculate_final_params/final_fn.pkl")

for nw in network_asc:
    params = reader.data_read_pkl(f"results/calculate_final_params/final_{nw}.pkl")
    # store number of transition
    n_p.append(params[0]["n_part"])
    n_f.append(params[0]["n_full"])
    n_p_s.append(np.sqrt(params[0]["n_part"]))
    n_f_s.append(np.sqrt(params[0]["n_full"]))

    # store transition time
    t_p_s.append(np.std(params[1]["t_t"]) / np.sqrt(len(params[1]["t_t"])))
    t_f_s.append(np.std(params[2]["t_t"]) / np.sqrt(len(params[2]["t_t"])))

    t_p.append(np.mean(params[1]["t_t"]))
    _, mw = mannwhitneyu(fn_data[1]["t_t"], params[1]["t_t"])
    t_p_mw.append(mw)
    t_f.append(np.mean(params[2]["t_t"]))
    _, mw = mannwhitneyu(fn_data[2]["t_t"], params[2]["t_t"])
    t_f_mw.append(mw)

    # store first passage time
    f_p_s.append(np.std(params[1]["fp_t"]) / np.sqrt(len(params[1]["fp_t"])))
    f_f_s.append(np.std(params[2]["fp_t"]) / np.sqrt(len(params[2]["fp_t"])))

    f_p.append(np.mean(params[1]["fp_t"]))
    _, mw = mannwhitneyu(fn_data[1]["fp_t"], params[1]["fp_t"])
    f_p_mw.append(mw)
    f_f.append(np.mean(params[2]["fp_t"]))
    _, mw = mannwhitneyu(fn_data[2]["fp_t"], params[2]["fp_t"])
    f_f_mw.append(mw)

    # Store sync time
    s_p_s.append(np.std(params[1]["s_t"]) / np.sqrt(len(params[1]["s_t"])))
    s_f_s.append(np.std(params[2]["s_t"]) / np.sqrt(len(params[2]["s_t"])))

    s_p.append(np.mean(params[1]["s_t"]))
    _, mw = mannwhitneyu(fn_data[1]["s_t"], params[1]["s_t"])
    s_p_mw.append(mw)
    s_f.append(np.mean(params[2]["s_t"]))
    _, mw = mannwhitneyu(fn_data[2]["s_t"], params[2]["s_t"])
    s_f_mw.append(mw)

    # store coverage
    # removing zero coverage values from partial transitions as they might seep due to very small values
    cov_part=params[1]["cov"]
    cov_part=[x for x in cov_part if x>0]
    cov_p_s.append(np.std(cov_part) / np.sqrt(len(cov_part)))
    cov_f_s.append(np.std(params[2]["cov"]) / np.sqrt(len(params[2]["cov"])))

    cov_p.append(np.mean(cov_part))
    _, mw = mannwhitneyu(fn_data[1]["cov"], cov_part)
    cov_p_mw.append(mw)
    cov_f.append(np.mean(params[2]["cov"]))
    _, mw = mannwhitneyu(fn_data[2]["cov"], params[2]["cov"])
    cov_f_mw.append(mw)

num_trans = {"n_p": n_p, "n_f": n_f, "n_p_s": n_p_s, "n_f_s": n_f_s}
trans_time = {
    "t_p": t_p,
    "t_f": t_f,
    "t_p_s": t_p_s,
    "t_f_s": t_f_s,
    "t_p_mw": t_p_mw,
    "t_f_mw": t_f_mw,
}
fp_time = {
    "f_p": f_p,
    "f_f": f_f,
    "f_p_s": f_p_s,
    "f_f_s": f_f_s,
    "f_p_mw": f_p_mw,
    "f_f_mw": f_f_mw,
}

sync_time = {
    "s_p": s_p,
    "s_f": s_f,
    "s_p_s": s_p_s,
    "s_f_s": s_f_s,
    "s_p_mw": s_p_mw,
    "s_f_mw": s_f_mw,
}

cov = {
    "cov_p": cov_p,
    "cov_f": cov_f,
    "cov_p_s": cov_p_s,
    "cov_f_s": cov_f_s,
    "cov_p_mw": cov_p_mw,
    "cov_f_mw": cov_f_mw,
}
```

```python
def fig_3_scatter(ax, parameter, ang_f, ang_p):

    full_vals = df_sc[(f"{parameter}", "full")]
    part_vals = df_sc[(f"{parameter}", "part")]

    x = df_bc["BET_CENT"]

    # --- Scatter plots ---
    ax.scatter(x, full_vals, c="purple", s=3, alpha=0.4)
    ax.scatter(x, part_vals, c="orange", s=3, alpha=0.4)

    # --- Full regression ---
    slope, intercept, _, _, _ = linregress(x, full_vals)
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = slope * x_line + intercept
    ax.plot(
        x_line, y_line, color="purple", linestyle="--", label="Full fit", linewidth=0.5
    )

    # Spearman correlation
    rho, pval = spearmanr(x, full_vals)
    mid_idx = len(x_line) // 2
    x_mid, y_mid = x_line[mid_idx], y_line[mid_idx]
    angle = np.degrees(np.arctan(slope)) + ang_f
    ax.text(
        x_mid,
        y_mid + 0.05 * (full_vals.max() - full_vals.min()),
        f"ρ={rho:.2f}, p={pval:.2g}",
        fontsize=4,
        color="purple",
        ha="center",
        rotation=angle,
        rotation_mode="anchor",
    )

    # --- Part regression ---
    slope, intercept, _, _, _ = linregress(x, part_vals)
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = slope * x_line + intercept
    ax.plot(
        x_line, y_line, color="orange", linestyle="--", label="Part fit", linewidth=0.5
    )

    # Spearman correlation
    rho, pval = spearmanr(x, part_vals)
    mid_idx = len(x_line) // 2
    x_mid, y_mid = x_line[mid_idx], y_line[mid_idx]
    angle = np.degrees(np.arctan(slope)) + ang_p
    ax.text(
        x_mid,
        y_mid + 0.05 * (part_vals.max() - part_vals.min()),
        f"ρ={rho:.2f}, p={pval:.2g}",
        fontsize=4,
        color="orange",
        ha="center",
        rotation=angle,
        rotation_mode="anchor",
    )

    # --- Labels and layout ---
    # ax.set_xlabel("BET_CENT")
    # ax.set_ylabel(f"{parameter}")
    # ax.legend()
```

Intepretation:  
- Steep slope implies faster transition  
- Slow slope implies slower transition  
- the point where slope changes is classifying that transitions which take time less and above that are different
- In Part & Full fpt, we can fit 2 and it seems feasible

Use below function when you also need uncertainties, which are helpful in calculating significance

```python
def fit_piecewiseLinear(data, ax, label):
    intervals = np.sort(data.copy())
    n = len(intervals)

    ccdf = 1.0 - np.arange(1, n + 1) / n
    mask = ccdf > 0
    intervals = intervals[mask]
    ccdf = ccdf[mask]

    log_ccdf = np.log(ccdf)
    x = intervals
    y = log_ccdf

    # Fit piecewise linear model with 2 segments
    my_pwlf = pwlf.PiecewiseLinFit(x, y)
    res = my_pwlf.fit(2)   # returns breakpoints

    # Predict
    y_hat = my_pwlf.predict(x)

    # Plot
    ax.scatter(x, y, label="Data")
    ax.plot(x, y_hat, 'r', label="Piecewise fit")
    ax.set_ylabel("log P(T > t)")
    ax.set_title(f"Log-CCDF breakpoint ~ {round(res[1])} {label}")
    ax.legend()

    # Slopes
    slopes = my_pwlf.slopes

    # --- Uncertainty estimation ---
    # Fit again with covariance output
    beta, cov = my_pwlf.fit(2, return_cov=True)

    # Standard errors
    se = np.sqrt(np.diag(cov))

    # Slopes uncertainties (first entries)
    slope_errors = se[:len(slopes)]

    # Breakpoint uncertainties (next entries)
    breakpoint_errors = se[len(slopes):]

    print("Slopes:", slopes)
    print("Slope SE:", slope_errors)
    print("Breakpoints:", res)
    print("Breakpoint SE:", breakpoint_errors)

    return {
        'slopes': slopes,
        'slope_uncertainties': slope_errors.tolist(),
        'breakpoints': res.tolist(),
        'breakpoint_uncertainties': breakpoint_errors.tolist()
    }
```

```python
def fit_piecewiseLinear(data, ax, label):
    # intervals as this is the time when state changes from another state
    
    intervals = data.copy()
    # print(f'original values: {intervals}')
    intervals = np.sort(intervals)
    n = len(intervals)

    # CCDF: P(T > t)
    ccdf = 1.0 - np.arange(1, n + 1) / n

    # Remove zero values (log undefined)
    mask = ccdf > 0
    intervals = intervals[mask]
    ccdf = ccdf[mask]

    # Log of CCDF
    log_ccdf = np.log(ccdf)

    x = intervals
    y = log_ccdf

    # Fit piecewise linear model with 2 segments
    my_pwlf = pwlf.PiecewiseLinFit(x, y)
    res = my_pwlf.fit(2)

    # Breakpoints returned by pwlf.fit
    breakpoints = res   # e.g. [min_x, breakpoint, max_x]

    # First segment is between breakpoints[0] and breakpoints[1]
    # finding out interval values which contributed to slope 1
    mask_slope1 = (x >= breakpoints[0]) & (x <= breakpoints[1])
    entries_slope1 = x[mask_slope1]

    # print("Entries belonging to slope 1:", entries_slope1.tolist())
     # Entries corresponding to slope 2 (between res[1] and res[2])
    mask2 = (x >= res[1]) & (x <= res[2])
    entries_slope2 = x[mask2]

    # Predict
    y_hat = my_pwlf.predict(x)

    # Plot
    ax.scatter(x, y, label="Data")
    ax.plot(x, y_hat, 'r', label="Piecewise fit")
    ax.set_ylabel("log P(T > t)")
    ax.set_title(f"Log-CCDF breakpoint ~ {round(res[1])} {label}")
    ax.legend()

    # Slopes of the fitted lines
    slopes = my_pwlf.slopes
    print("Slopes of fitted lines:", slopes)

    # Optionally annotate slopes on the plot
    for i, slope in enumerate(slopes):
        ax.text(x.mean(), y_hat.mean() - i*0.5, f"Slope {i+1}: {slope:.3f}", color="blue")

    return {'exp':slopes, 'joint': round(res[1]),'slope_1_entries':entries_slope1.tolist(),'s2':entries_slope2.tolist()}
```

```python
def scatter_plot_btwn_part_full(avg_comm_per, network, ax=None): 
    # Sort 'Full Transitions' by value
    full_sorted = dict(sorted(avg_comm_per['avg_full'].items(), key=lambda x: x[1]))

    # Reorder 'Partial Transitions' to match the order of 'Full Transitions' keys
    part_sorted = {k: avg_comm_per['avg_part'].get(k, 0) for k in full_sorted.keys()}
    print(full_sorted)
    print(part_sorted)

    # Use provided axis or fallback to current axis
    if ax is None:
        ax = plt.gca()

    # Create scatter plot on the given axis
    ax.scatter(full_sorted.values(), part_sorted.values(), color='orange')

    # Add labels for each point
    for k, x, y in zip(full_sorted.keys(), full_sorted.values(), part_sorted.values()):
        ax.text(x, y, k, fontsize=9)

    ax.set_xlabel('Full Transitions (%)')
    ax.set_ylabel('Partial Transitions (%)')
    ax.set_title(f'Scatter Plot: Full vs Partial Transitions {network}')
    ax.grid(True)

```
