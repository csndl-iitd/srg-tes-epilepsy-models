
### Calculate the parameters Transition time, Sync time, FPT , Coverage using NC data and number of transitions

#### Data folder:  
- data/raw/workstation for Global Order, Local Order and Node Community data
- data/raw/workstation/coverage_nc for Coverage data of any network (this uses compute_coverage_nc() function in commAnalysis.py)
- For above data, the thresholds are at results/coverage_threshold.pkl

#### Import necessary libraries

```python
import numpy as np
import pandas as pd
from tes.data_ingestion import DataRead
from tes.analysis import AnalysisFunc
from tes.algo_umap import UmapAlgo
from tes.algo_partial_full import AlgoParFull
from tes.commAnalysis import CommPercentages
from tes.DataUtils import DataUtils
from scipy.stats import mannwhitneyu
from scipy.stats import spearmanr, linregress
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances_argmin
import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt
from matplotlib import cm
from pathlib import Path
import os
import glob
import networkx as nx
import pwlf
import pickle
import umap
```

#### Make objects to use the functions

```python
algo_umap = UmapAlgo()
algo= AlgoParFull()
analyse= AnalysisFunc()
reader = DataRead()
comm_ops = CommPercentages()
utils = DataUtils()
```

#### Read all data i.e  global order and node community data
- stored in data/raw/workstation 

```python
data_folder = Path('../../data/raw/workstation')
files_go=glob.glob(str(data_folder/'tes_*_go.bz2'))
files_nc=glob.glob(str(data_folder / 'tes_*_nc.bz2'))
```

#### Read Global Order and Node community files and compute all parameters separately for partial and full transitions

```python
networks = list(reader.data_read_pkl('results/coverage_threshold.pkl').keys())
```

```python
# umcomment if you want to calculate params once again
for f_go, f_nc, nw in zip(files_go,files_nc, networks):
    # read files
    df_go = pd.read_pickle(f_go,compression='bz2')
    df_nc = pd.read_pickle(f_nc,compression='bz2')
    data = algo.compute_final_param_df(df_go,df_nc,nw)
    reader.data_dump_pkl(f'results/calculate_final_params/final_{nw}.pkl',data)
```

Above results are stored in folder '../../data/results/calculate_final_params/final_{nw}.pkl'  
- nw = ['bla', 'enti', 'fn', 'lha', 'lsc', 'mob', 'mop', 'mos', 'mrn', 'orb', 'peri', 'pir', 're']  
- It is stored as a pickle file which is actually a tuple  
- Tuple:  
index 0: dictionary with number of partial, full and total transition and iteration numbers with transitions  
        KEYS: ['n_part', 'n_full', 'n_total', 'itr_tran']  
index 1: partial transitions - KEY: ['t_t', 'fp_t', 's_t', 'cov']  
index 2: full transitions - KEYS: ['t_t', 'fp_t', 's_t', 'cov']
