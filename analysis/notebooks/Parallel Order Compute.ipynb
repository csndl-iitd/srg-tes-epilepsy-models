# Parallel computation for global and  local order

Follow the template for parallel execution, though requires large memory

```python
from tes.data_ingestion import DataRead
from concurrent.futures import ProcessPoolExecutor, as_completed
```

```python
data=DataRead()
```

```python
simulations=[
    (1,'go_1.bz2','lo_1.bz2'),
    (2,'go_2.bz2','lo_2.bz2'),
    (3,'go_3.bz2','lo_3.bz2')
]

with ProcessPoolExecutor(max_workers=3) as executor:
    futures=[executor.submit(data.ext_global_order,itr_c, itr_filename, loc_filename) for itr_c, itr_filename, loc_filename in simulations]
    for future in as_completed(futures): 
        try: 
            future.result() 
        except Exception as e: 
            print(f"Simulation raised an exception: {e}")
```
