# Extract node community using local order data  
1. Extract node communities data    

```python
from tes.data_ingestion import DataRead
```

Create object to read data

```python
nc_ext=DataRead()
```

Store local order data in dataframe

```python
df=nc_ext.data_read_bz2('l20_fn.bz2')
```

#### 1. Extract node communities data  

```python
threshold = 0.3
nc_ext.ext_node_community(df,threshold,'nc3_fn.bz2')
```
